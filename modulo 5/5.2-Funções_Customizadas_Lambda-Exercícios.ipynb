### Q1.

- Sabendo que o código a seguir calcula o fatorial de n, escreva uma função chamada ```fatorial()``` que recebe um inteiro ```n``` como parâmetro e retorna o resultado do fatorial de ```n```. 
- No programa principal, peça ao usuário o valor de ```n```, chame a sua função e imprima o retorno.

```python
fat = 1
for i in range(1, n+1):
    fat *= i 
```

````python
def fatorial(n):
    if n ==0 or n == 1:
        return 1
    return n * fatorial(n-1)



````

### Q3.

- Crie uma função em Python chamada ```soma_digitos``` que recebe um número inteiro como parâmetro e retorna a soma dos seus dígitos. Por exemplo, para o número 123, a função deve retornar 6, $(1 + 2 + 3)$.
    - O desafio aqui é separar os dígitos de um número inteiro usando operações aritméticas <br> <br>

- No programa principal solicite ao usuário que insira um número e utilize a função ```soma_digitos``` para calcular e exibir a soma dos seus dígitos.

````python
def soma_digitos(n):
    soma = 0
    n = abs(n)
    while n>0:
        soma += n % 10
        n = n // 10
    return soma
````

````python
## Escreva aqui o programa principal e execute a célula
numero = int(input("Digite um número inteiro: "))
resultado = soma_digitos(numero)
print(f"A soma dos digitos de {numero} é: {resultado} ")
````

### Q4.

- Crie a função ```inverteValor()``` que recebe um inteiro de qualquer tamanho e retorna esse valor invertido usando apenas operações aritméticas
- Crie a função ```verificaInverso()``` que recebe o valor original e o valor invertido e retorna verdadeiro se ambos forem igualmente par ou igualmente ímpar. Retorne falso caso contrário.
- No programa principal, peça um valor do usuário e imprima o retorno de ambas as funções.


````python
## Crie aqui as funções inverteValor() e verificaInverso()
def inverteValor(n):
    inverso = 0
    while n > 0:
        digito = n % 10
        inverso = (inverso * 10) + digito
        n = n // 10
    return inverso

def verificaInverso(original, invertido):
    if (original % 2 == 0 and invertido % 2 == 0) or \
       (original % 2 != 0 and invertido % 2 != 0):
        return True
    return False
````

````python
## Escreva aqui o programa principal e execute a célula
valor = int(input("Digite um valor inteiro: "))
valor_invertido = inverteValor (valor)
resultado_verificacao = verificaInverso(valor, valor_invertido)

print(f"Valor invertido: {valor_invertido}")
print(f"Mesma paridade: {resultado_verificacao}")
````

````python
## Crie aqui as funções 
import math

def calcula_perimetro_triangulo(a, b, c):
    return a + b + c

def calcula_perimetro_circulo(raio):
    return 2 * math.pi * raio

def calcula_perimetro_retangulo(lado1, lado2=None):
    if lado2 is None:
        return 4 * lado1
    return 2 * lado1 + 2 * lado2


````

````python
## Escreva aqui o programa principal 
def menu():
    while True:
        print("\n--- Sistema de Geometria ---")
        print("1. Perimetro do triangulo")
        print("2. Perimetro do circulo")
        print("3. Perimetro do retangulo/quadrado")
        print("4. Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == '1':
            l1 = int(input("Lado 1: "))
            l2 = int(input("Lado 2: "))
            l3 = int(input("Lado 3: "))
            print(f"Resultado: {calcula_perimetro_triangulo(l1, l2, l3)}")

        elif opcao == '2':
            r = int(input("Raio de circulo: "))
            print(f"Resultado: {calcula_perimetro_circulo(r):.2f}")

        elif opcao == '3':
            l1 = int(input("Lado 1: " ))
            entrada_l2 = input("Lado 2 (vazio para quadrado): ")
            if entrada_l2 == "":
                print(f"Resultado (quadrado):{calcula_perimetro_retangulo (l1)}")
            else:
                l2 = int(entrada_l2)
                print (f"Resultado (retangulo): {calcula_perimetro_retangulo (l1, l2)}")

        elif opcao == '4':
            print("Encerrando programa...")
            break
        else:
            print("Opção inválida!")
if __name__=="__main__":
    menu()
````

### Q6.

Crie uma função lambda para verificar se um número é par ou ímpar. Em seguida, solicite ao usuário um número indefinidos de valores (até que o usuário digite 0). Para cada valor de entrada, informe se é par ou ímpar.

Exemplo de interação:
```
Digite os valores que deseja verificar a paridade (digite 0 para finalizar a entrada de dados):
3
ímpar
8
par
12
par
5
ímpar
7ímpar
```

````python
verificar_paridade = lambda n: "par" if n % 2 == 0 else "ímpar"

while True:
    try:
        valor = int(input("digite os valores que deseja verificar a paridade (digite 0para finalizar a entrada de dados): "))

        if valor == 0:
            break

        resultado = verificar_paridade(valor)
        print(f"{valor} é {resultado}")

    except ValueError:
        print("Por favor, digite apenas numeros inteiros.")
        
````

**Sua solução deve incluir pelo menos uma função ```lambda```**


````python
## Escreva aqui o programa principal 
print("Opções: (1) maior ou (2) menor?" )
opcao = input("Opção: ")

if opcao == '1':
    operacao = lambda lista: max(lista)
    rotulo = "maior"
else:
    operacao = lambda lista: min(lista)
    rotulo = "menor"

valores = []
print("\nDigite os valoresde entrada. Digite 0para finalizara entrada de valores.")

while True:
    valor = float(input())
    if valor == 0:
        break
    valores.append(valor)

if valores:
    resultado = operacao(valores)
    print(f"\n0 {rotulo} valor digitado foi: {resultado}")

else:
    print("\nNenhum valor foi inserido.")
    
````
